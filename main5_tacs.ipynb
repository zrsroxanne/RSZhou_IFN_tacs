```python
# S6 supplement: pairwise dataset correlations of AD-NC different surface area across 18 networks
# Each point is one Yeo network; coordinates are exported for every panel.
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
import seaborn as sns
from matplotlib import font_manager
from scipy import stats

S6_AREA_CORR_DIR = Path(
    '/data/home/rszhou/xwtacs_folder_zhou/reviese_result/'
    'S6/S6_0_mcad_adni_oasis_surface_area_correlations'
)
S6_AREA_CORR_DIR.mkdir(parents=True, exist_ok=True)

required_area_vars_s6 = ['mcad_diff_area', 'adni_diff_area', 'oasis_diff_area']
missing_area_vars_s6 = [
    name for name in required_area_vars_s6 if name not in globals()
]
if missing_area_vars_s6:
    raise NameError(
        f'Missing {missing_area_vars_s6}. Run show_average_mcad for MCAD, '
        'ADNI, and OASIS first.'
    )


def s6_standardize_area_table(area_df, dataset_name):
    out = pd.DataFrame(area_df).copy()
    if not {'Network', 'Area'}.issubset(out.columns):
        raise KeyError(
            f'{dataset_name} area table must contain Network and Area; '
            f'found {out.columns.tolist()}'
        )
    out['network_id'] = pd.to_numeric(out['Network'], errors='coerce')
    out['surface_area'] = pd.to_numeric(out['Area'], errors='coerce')
    out = out.dropna(subset=['network_id', 'surface_area']).copy()
    out['network_id'] = out['network_id'].astype(int)
    if out['network_id'].duplicated().any():
        raise ValueError(f'{dataset_name} contains duplicated network IDs.')
    expected = set(range(1, 19))
    observed = set(out['network_id'])
    if observed != expected:
        raise ValueError(
            f'{dataset_name} network IDs must be 1..18; '
            f'missing={sorted(expected-observed)}, extra={sorted(observed-expected)}'
        )
    return (
        out[['network_id', 'surface_area']]
        .sort_values('network_id')
        .reset_index(drop=True)
    )


area_tables_s6 = {
    'MCAD': s6_standardize_area_table(mcad_diff_area, 'MCAD'),
    'ADNI': s6_standardize_area_table(adni_diff_area, 'ADNI'),
    'OASIS': s6_standardize_area_table(oasis_diff_area, 'OASIS'),
}

if 'NET_NAMES' in globals() and len(NET_NAMES) >= 18:
    area_network_names_s6 = [str(x) for x in NET_NAMES[:18]]
elif 'netnames' in globals() and len(netnames) >= 18:
    area_network_names_s6 = [str(x) for x in netnames[:18]]
else:
    area_network_names_s6 = [f'Net{i}' for i in range(1, 19)]

if 'yeo_color_array_18_no_ven' in globals():
    area_network_colors_s6 = np.asarray(yeo_color_array_18_no_ven)[:18]
else:
    area_network_colors_s6 = plt.cm.tab20(np.linspace(0, 1, 18))
area_network_hex_s6 = [
    mpl.colors.to_hex(color) for color in area_network_colors_s6
]

pair_specs_s6 = [
    ('MCAD', 'ADNI'),
    ('MCAD', 'OASIS'),
    ('ADNI', 'OASIS'),
]

coordinate_tables_s6 = []
correlation_rows_s6 = []
for pair_index_s6, (x_dataset_s6, y_dataset_s6) in enumerate(pair_specs_s6):
    coord_s6 = area_tables_s6[x_dataset_s6].rename(
        columns={'surface_area': 'x_surface_area'}
    ).merge(
        area_tables_s6[y_dataset_s6].rename(
            columns={'surface_area': 'y_surface_area'}
        ),
        on='network_id',
        how='inner',
        validate='one_to_one',
    )
    coord_s6['network_name'] = [
        area_network_names_s6[i - 1] for i in coord_s6['network_id']
    ]
    coord_s6['point_color'] = [
        area_network_hex_s6[i - 1] for i in coord_s6['network_id']
    ]
    coord_s6['comparison'] = f'{x_dataset_s6}_vs_{y_dataset_s6}'
    coord_s6['x_dataset'] = x_dataset_s6
    coord_s6['y_dataset'] = y_dataset_s6
    coord_s6['metric'] = 'AD-NC different surface area'
    coord_s6 = coord_s6[
        [
            'comparison', 'x_dataset', 'y_dataset',
            'network_id', 'network_name',
            'x_surface_area', 'y_surface_area',
            'point_color', 'metric',
        ]
    ]
    coordinate_tables_s6.append(coord_s6)

    x_s6 = coord_s6['x_surface_area'].to_numpy(dtype=float)
    y_s6 = coord_s6['y_surface_area'].to_numpy(dtype=float)
    pearson_r_s6, pearson_p_s6 = stats.pearsonr(x_s6, y_s6)
    spearman_r_s6, spearman_p_s6 = stats.spearmanr(x_s6, y_s6)
    regression_s6 = stats.linregress(x_s6, y_s6)
    correlation_rows_s6.append({
        'comparison': f'{x_dataset_s6}_vs_{y_dataset_s6}',
        'x_dataset': x_dataset_s6,
        'y_dataset': y_dataset_s6,
        'metric': 'AD-NC different surface area',
        'n_networks': int(len(coord_s6)),
        'pearson_r': float(pearson_r_s6),
        'pearson_p': float(pearson_p_s6),
        'spearman_rho': float(spearman_r_s6),
        'spearman_p': float(spearman_p_s6),
        'regression_slope': float(regression_s6.slope),
        'regression_intercept': float(regression_s6.intercept),
        'regression_r_squared': float(regression_s6.rvalue ** 2),
        'regression_slope_p': float(regression_s6.pvalue),
        'regression_slope_se': float(regression_s6.stderr),
    })

s6_area_scatter_coordinates = pd.concat(
    coordinate_tables_s6, ignore_index=True
)
s6_area_pairwise_correlations = pd.DataFrame(correlation_rows_s6)

all_coordinates_csv_s6 = (
    S6_AREA_CORR_DIR /
    'S6_MCAD_ADNI_OASIS_pairwise_surface_area_scatter_coordinates.csv'
)
stats_csv_s6 = (
    S6_AREA_CORR_DIR /
    'S6_MCAD_ADNI_OASIS_pairwise_surface_area_correlations.csv'
)
s6_area_scatter_coordinates.to_csv(all_coordinates_csv_s6, index=False)
s6_area_pairwise_correlations.to_csv(stats_csv_s6, index=False)

for coord_s6 in coordinate_tables_s6:
    comparison_s6 = coord_s6['comparison'].iloc[0]
    coord_s6.to_csv(
        S6_AREA_CORR_DIR /
        f'S6_{comparison_s6}_surface_area_scatter_coordinates.csv',
        index=False,
    )


def s6_format_p(p_value):
    if p_value < 0.001:
        return 'p < 0.001'
    return f'p = {p_value:.3f}'


available_fonts_s6 = {f.name for f in font_manager.fontManager.ttflist}
font_family_s6 = next(
    (f for f in ['Arial', 'Liberation Sans', 'Nimbus Sans', 'DejaVu Sans']
     if f in available_fonts_s6),
    'DejaVu Sans',
)
plt.rcParams.update({
    'font.family': 'sans-serif',
    'font.sans-serif': [font_family_s6],
    'font.size': 7,
    'axes.linewidth': 0.7,
    'xtick.major.width': 0.7,
    'ytick.major.width': 0.7,
    'pdf.fonttype': 42,
    'ps.fonttype': 42,
})

fig, axes = plt.subplots(1, 3, figsize=(8.7, 2.65))
for pair_index_s6, (
    ax_s6, coord_s6, (x_dataset_s6, y_dataset_s6)
) in enumerate(zip(axes, coordinate_tables_s6, pair_specs_s6)):
    x_s6 = coord_s6['x_surface_area'].to_numpy(dtype=float)
    y_s6 = coord_s6['y_surface_area'].to_numpy(dtype=float)

    # Regression line and 95% bootstrap confidence band.
    sns.regplot(
        data=coord_s6,
        x='x_surface_area',
        y='y_surface_area',
        scatter=False,
        ci=95,
        n_boot=5000,
        seed=20260729 + pair_index_s6,
        color='#7F7F7F',
        line_kws={'linewidth': 1.2},
        ax=ax_s6,
    )
    ax_s6.scatter(
        x_s6,
        y_s6,
        s=19,
        c=coord_s6['point_color'].tolist(),
        edgecolors='white',
        linewidths=0.25,
        alpha=0.95,
        zorder=3,
    )

    stat_row_s6 = s6_area_pairwise_correlations[
        s6_area_pairwise_correlations['comparison']
        == f'{x_dataset_s6}_vs_{y_dataset_s6}'
    ].iloc[0]
    annotation_s6 = (
        f"r = {stat_row_s6['pearson_r']:.2f}, "
        f"{s6_format_p(stat_row_s6['pearson_p'])}"
    )
    ax_s6.text(
        0.05, 0.82,
        annotation_s6,
        transform=ax_s6.transAxes,
        ha='left',
        va='top',
        fontsize=7.2,
        color='black',
    )

    x_span_s6 = float(np.nanmax(x_s6) - np.nanmin(x_s6))
    y_span_s6 = float(np.nanmax(y_s6) - np.nanmin(y_s6))
    x_pad_s6 = 0.06 * x_span_s6 if x_span_s6 > 0 else 1.0
    y_pad_s6 = 0.06 * y_span_s6 if y_span_s6 > 0 else 1.0
    ax_s6.set_xlim(
        min(0.0, float(np.nanmin(x_s6)) - x_pad_s6),
        float(np.nanmax(x_s6)) + x_pad_s6,
    )
    ax_s6.set_ylim(
        min(0.0, float(np.nanmin(y_s6)) - y_pad_s6),
        float(np.nanmax(y_s6)) + y_pad_s6,
    )
    ax_s6.set_xlabel(f'{x_dataset_s6} AD-NC different area', fontsize=7.5)
    ax_s6.set_ylabel(f'{y_dataset_s6} AD-NC different area', fontsize=7.5)
    ax_s6.spines['top'].set_visible(False)
    ax_s6.spines['right'].set_visible(False)
    ax_s6.tick_params(length=2.6, pad=1.5, labelsize=6.5)

fig.tight_layout(w_pad=1.1)
fig_png_s6 = (
    S6_AREA_CORR_DIR /
    'S6_MCAD_ADNI_OASIS_pairwise_surface_area_correlations.png'
)
fig_pdf_s6 = (
    S6_AREA_CORR_DIR /
    'S6_MCAD_ADNI_OASIS_pairwise_surface_area_correlations.pdf'
)
fig.savefig(fig_png_s6, dpi=600, bbox_inches='tight')
fig.savefig(fig_pdf_s6, bbox_inches='tight')
plt.show()

print('Saved correlation statistics:', stats_csv_s6)
print('Saved all scatter coordinates:', all_coordinates_csv_s6)
print('Saved figure:', fig_png_s6)
display(s6_area_pairwise_correlations)
display(s6_area_scatter_coordinates)
```

```python
# MCAD subject-level network weighted cortical surface area and NC/MCI/AD comparison
from pathlib import Path
import numpy as np
import pandas as pd
import nibabel as nib

S6_FS4_TEMPLATE_DIR = Path(
    '/data/home/rszhou/xwtacs_folder_zhou/Pin_Work/nits/templates/'
    'fsaverage/fsaverage4'
)
s6_lh_area_path = (
    S6_FS4_TEMPLATE_DIR /
    'tpl-fsaverage_den-3k_hemi-L_pial_desc-area.shape.gii'
)
s6_rh_area_path = (
    S6_FS4_TEMPLATE_DIR /
    'tpl-fsaverage_den-3k_hemi-R_pial_desc-area.shape.gii'
)
if not s6_lh_area_path.exists() or not s6_rh_area_path.exists():
    raise FileNotFoundError(
        f'Cannot find fsaverage4 area files: '
        f'{s6_lh_area_path}, {s6_rh_area_path}'
    )

# Vertex order in mcad_indipar is LH first (0:2562), then RH (2562:5124).
s6_lh_area_weight = np.asarray(
    nib.load(str(s6_lh_area_path)).agg_data(), dtype=float
).reshape(-1)
s6_rh_area_weight = np.asarray(
    nib.load(str(s6_rh_area_path)).agg_data(), dtype=float
).reshape(-1)
if s6_lh_area_weight.shape != (2562,) or s6_rh_area_weight.shape != (2562,):
    raise ValueError(
        f'Expected 2562 area weights per hemisphere, got '
        f'LH={s6_lh_area_weight.shape}, RH={s6_rh_area_weight.shape}.'
    )
s6_fs4_area_weight_correct_order = np.concatenate([
    s6_lh_area_weight,
    s6_rh_area_weight,
])

s6_mcad_indipar_array = np.asarray(mcad_indipar)
if (
    s6_mcad_indipar_array.ndim != 3
    or s6_mcad_indipar_array.shape[2] != 5124
):
    raise ValueError(
        f'Expected mcad_indipar subject x network x 5124, '
        f'got {s6_mcad_indipar_array.shape}.'
    )
if s6_mcad_indipar_array.shape[0] != len(mcad_df):
    raise ValueError(
        f'Subject mismatch: mcad_indipar={s6_mcad_indipar_array.shape[0]}, '
        f'mcad_df={len(mcad_df)}.'
    )

# Hard individualized-network label for every subject and vertex.
s6_mcad_hard_labels = s6_mcad_indipar_array.argmax(axis=1)

# Raw subject-level weighted cortical surface area for networks 1..18.
mcad_weighted_surface_area = np.column_stack([
    np.sum(
        (s6_mcad_hard_labels == network_id)
        * s6_fs4_area_weight_correct_order[None, :],
        axis=1,
    )
    for network_id in range(1, 19)
])

# Keep mcad_ares as a compatibility alias, but it now contains weighted mm2,
# not mcad_indipar.sum(axis=2).
mcad_ares = mcad_weighted_surface_area

# Save the raw subject-level values in long format for audit and table export.
s6_weighted_area_raw_wide = mcad_df[
    ['sub', 'Group']
].reset_index(drop=True).copy()
for network_id in range(1, 19):
    s6_weighted_area_raw_wide[str(network_id)] = (
        mcad_weighted_surface_area[:, network_id - 1]
    )
s6_weighted_area_raw_long = s6_weighted_area_raw_wide.melt(
    id_vars=['sub', 'Group'],
    value_vars=[str(i) for i in range(1, 19)],
    var_name='Network',
    value_name='Raw_weighted_surface_area_mm2',
)

df = show_area_boxplot2(
    mcad_ares,
    mcad_df,
    indx=None,
    save_path=(
        '/data/home/rszhou/xwtacs_folder_zhou/reviese_result/'
        'S6/Network_weighted_surface_area_compare.pdf'
    ),
    y_label='Weighted cortical surface area (mm²)',
)

print(
    'mcad_ares now contains raw subject-level fsaverage4 weighted '
    'surface area for 18 individualized networks.'
)
print('Shape:', mcad_ares.shape)
```
